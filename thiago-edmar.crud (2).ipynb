```python
class Cliente():
    id_cliente:int
    nome: str
    telofone: str 
    email: str

    def __init__(self, nome: str, telefone: str | None, email: str | None, id_cliente: int | None = None):
            self.id_cliente = id_cliente
            self.nome = nome
            self.telefone = telefone
            self.email = email

    def __str__(self):
            return f"Cliente(id_cliente={self.id_cliente}, nome={self.nome}, telefone={self.telefone}, email={self.email})"

    def __repr__(self):
            return f"Cliente({self.id_cliente} - {self.nome})"
```

```python
def criar_cliente(nome: str, telefone: str | None, email: str | None) -> Cliente | None:
        try:
            cursor.execute(
                "INSERT INTO cliente (nome, telefone, email) "
                "VALUES (%s, %s, %s) RETURNING id_cliente",
                (nome, telefone, email)
            )
            id_cliente = cursor.fetchone()[0]  # o id que o SERIAL acabou de gerar
            conexao.commit()
            return Cliente(nome, telefone, email, id_cliente)
        except pg.DatabaseError as e:
            print(e)
            conexao.rollback()
        return None
```

```python
def listar_clientes() -> list[Cliente]:
        clientes = []
        try:
            cursor.execute("SELECT id_cliente, nome, telefone, email FROM cliente ORDER BY id_cliente")
            for linha in cursor.fetchall():
                clientes.append(Cliente(linha[1], linha[2], linha[3], linha[0]))
        except pg.DatabaseError as e:
            print(e)
            conexao.rollback()
        return clientes
```

```python
def buscar_cliente(id_cliente: int) -> Cliente | None:
        try:
            cursor.execute(
                "SELECT id_cliente, nome, telefone, email FROM cliente WHERE id_cliente = %s",
                (id_cliente,)  # a vírgula faz disso uma tupla
            )
            linha = cursor.fetchone()  # None se nenhuma linha bateu com o WHERE
            if linha:
                return Cliente(linha[1], linha[2], linha[3], linha[0])
        except pg.DatabaseError as e:
            print(e)
            conexao.rollback()
        return None
```

```python
def atualizar_cliente(cliente: Cliente) -> bool:
        try:
            cursor.execute(
                "UPDATE cliente SET nome = %s, telefone = %s, email = %s WHERE id_cliente = %s",
                (cliente.nome, cliente.telefone, cliente.email, cliente.id_cliente)
            )
            alteradas = cursor.rowcount  # 0 = não existe cliente com esse id
            conexao.commit()
            return alteradas > 0
        except pg.DatabaseError as e:
            print(e)
            conexao.rollback()
        return False
```

```python
def deletar_cliente(id_cliente: int) -> bool:
        try:
            cursor.execute("DELETE FROM cliente WHERE id_cliente = %s", (id_cliente,))
            apagadas = cursor.rowcount
            conexao.commit()
            return apagadas > 0
        except pg.DatabaseError as e:
            print(e)  # ex.: o cliente tem agendamento, e a chave estrangeira impede o DELETE
            conexao.rollback()
        return False
```

```python
class Quadra():
    id_quadra:int           
    preco: float 
    coberta: bool  
    
    def __init__(self, id_quadra: int, preco: float, coberta: bool):
            self.id_quadra= id_quadra
            self.preco= preco
            self.coberta = coberta

    def __str__(self):
            return f"Quadra(id_quadra={self.id_quadra}, preco={self.preco}, coberta={self.coberta})"

    def __repr__(self):
            return f"Quadra({self.id_quadra} - {self.preco})"
```

```python
def criar_quadra(preco: float, coberta:bool | None) -> Quadra | None:
        try:
            cursor.execute(
                "INSERT INTO cliente (preco, coberta) "
                "VALUES (%s, %s) RETURNING id_quadra",
                (preco, coberta)
            )
            id_quadra = cursor.fetchone()[0]  # o id que o SERIAL acabou de gerar
            conexao.commit()
            return Quadra(preco, coberta, id_quadra)
        except pg.DatabaseError as e:
            print(e)
            conexao.rollback()
        return None
```

```python
def listar_quadras() -> list[Quadra]:
        quadras = []
        try:
            cursor.execute("SELECT id_quadra, preco, coberta, FROM quadra ORDER BY id_quadra")
            for linha in cursor.fetchall():
                quadras.append(Quadra(linha[1], linha[2], linha[0]))
        except pg.DatabaseError as e:
            print(e)
            conexao.rollback()
        return quadras
```

```python
def buscar_quadra(id_quadra: int) -> Quadra | None:
        try:
            cursor.execute(
                "SELECT id_quadra, preco, coberta FROM quadra WHERE id_quadra = %s",
                (id_quadra,)  # a vírgula faz disso uma tupla
            )
            linha = cursor.fetchone()  # None se nenhuma linha bateu com o WHERE
            if linha:
                return Cliente(linha[1], linha[2], linha[3], linha[0])
        except pg.DatabaseError as e:
            print(e)
            conexao.rollback()
        return None
```

```python
def atualizar_quadra(Quadra: Quadra) -> bool:
        try:
            cursor.execute(
                "UPDATE quadra SET preco = %s, coberta = %s WHERE id_cliente = %s",
                (quadra.preco, quadra.coberta, quadra.id_quadra)
            )
            alteradas = cursor.rowcount  
            conexao.commit()
            return alteradas > 0
        except pg.DatabaseError as e:
            print(e)
            conexao.rollback()
        return False
```

```python
def deletar_quadra(id_quadra: int) -> bool:
        try:
            cursor.execute("DELETE FROM quadra WHERE id_quadra = %s", (id_quadra,))
            apagadas = cursor.rowcount
            conexao.commit()
            return apagadas > 0
        except pg.DatabaseError as e:
            print(e)  # ex.: o cliente tem agendamento, e a chave estrangeira impede o DELETE
            conexao.rollback()
        return False
```

```python
class Partida():
    id_partida: int
    nome_artilheiro: str
    total_gols: int

    def __init__(self, nome_artilheiro:str, total_gols:int):
        self.nome_artilheiro=nome_artilheiro
        self.total_gols=total_gols

    def __str__(self):
        return f"Partida( nome_artilheiro[{self.nome_artilheiro}, total_gols={self.total_gols}])"

    def __repr__(self):
        return f"Partida {self.nome_artilheiro}"
```
